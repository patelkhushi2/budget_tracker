```python
import sqlite3
import pandas as pd
conn = sqlite3.connect("budget.db")
```

```python
query = """
SELECT *
FROM budgets;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT *
FROM categories;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT categories.category_name, budgets.budget_amount
FROM budgets
JOIN categories
ON budgets.category_id = categories.category_id;
"""

budget_df = pd.read_sql_query(query, conn)

budget_df
```

```python
query = """
SELECT category_id, SUM(amount) AS total_spent
FROM transactions
WHERE transaction_type = 'Purchase'
GROUP BY category_id;
"""

spending_df = pd.read_sql_query(query, conn)

spending_df
```

```python
query = """
SELECT 
    categories.category_name,
    SUM(transactions.amount) AS total_spent
FROM transactions
JOIN categories
    ON transactions.category_id = categories.category_id
WHERE transactions.transaction_type = 'Purchase'
GROUP BY categories.category_name;
"""

spending_df = pd.read_sql_query(query, conn)

spending_df
```
